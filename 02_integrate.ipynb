### 2. Resolving schema conflicts
Column names and data types appear to be the main problem that needs resolution. Systematically reviewing the data set column by column I will attempt to resolve these conflicts.

#### Conflict 1: ClothID in df_orig and Clothing ID in df_2
These columns represent the same information. All that is required is to change the name of ClothID to match it's counterpart in in df2 (the schema which is being adopted). We also need to ensure that the data type of df_orig['ClothID'] matches that of df2.

```python
# Rename ClothID column to match df2
df_orig = df_orig.rename(columns = {'ClothID':'Clothing ID'})
df_orig.head()
```

```python
# Rename Review_Title to Title
df_orig = df_orig.rename(columns = {'Review Title':'Title'})
df_orig.head()
```

#### Conflict 3: Customer_Rating in df_orig vs Rating in df2
These columns represent the same information. Both columns already have the same data type. So we just have to rename the column in df_orig

```python
# Rename Customer_Rating to Rating
df_orig = df_orig.rename(columns = {'Customer Rating':'Rating'})
df_orig.head()
```

#### Conflict 4: Positive_Review_Count in df_orig vs Positive Feedback Count in df2
These columns represent the same information. Both columns already have the same data type. So we just have to rename the column in df_orig

```python
# Rename Positive Review Count to Positive Feedback Count
df_orig = df_orig.rename(columns = {'Positive Review Count':'Positive Feedback Count'})
df_orig.head()
```

#### Conflict 5: Section in df_orig vs Division Name in df2

### 4. Resolving data conflicts:
Because we have concatenated the data, we need to check for duplicates in the merged dataframe, where the same data was contained in both the original data sets.

#### Conflict 1: Check for Duplicates.
Now that the two datasets are merged, we will look for duplicates. Because there is no unique key (unique column that should uniquely identify each record), the first check is for exact duplicates: records that are identical in every column. Such records are almost certainly the same review appearing in both datasets, as it is near impossible for a specific clothing item to be reviewed by someone of the same age, giving their review with the exact same title, taking exactly the same time to post the review, AND paying the exact same price.

Code adapted from course materials on data integration [1]

```python
### check exact duplication - 342 results
duplicates = merged_df[merged_df.duplicated(keep=False)] # showing all duplicated records
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=duplicates.columns.tolist()).head(10)
```

342 records are involved in exact duplication, which is 171 pairs. One copy of each pair is dropped.

```python
# Code to fix the conflict - Can see now there are 13735 records after duplicates have been removed.
merged_df = merged_df.drop_duplicates(keep='last')
merged_df.shape
```

#### Finding global key for the data
For this data, it is difficult to obtain a global key, as the Clothing ID variable is not unique, and there are no unique identifiers for users. Therefore the global key will be a combination of different columns and should uniquely identify a particular record. Considering the type of data each variable holds, and testing various keys to see if they are suitable or not by checking for duplicates with specific combinations. The aim of the global key is to be able to identify any particular record uniquely from the key, and the key is minimal/contains the least number of attributes possible. In my opinion, it would be unwise to exclude Clothing ID as it provides an anchor to the specific clothing item being reviewed. So the combination of variables that make up the global key should include clothing ID.

### Check duplication for specific unique combination columns 

```python
# Code to identify global key
# check duplication for specific unique combination columns 
cols = ['Clothing ID'] # Clothing ID alone cannot be the global key as there are many duplicates identified using this as the key alone
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Clothing ID', 'Age'] # Clothing ID + Age cannot be the global key it is possible for a particular clothing item to be reviewed by multiple people of the same age.
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Clothing ID', 'Age', 'Title'] # Clothing ID + Age + Title cannot be the global key.
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Clothing ID', 'Price'] # Clothing ID and Price cannot be the global key
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Clothing ID', 'Active Time'] # Clothing ID and Time cannot be the global key
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Clothing ID', 'Price', 'Title'] # Clothing ID  + Price  + Title can be a global key
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

```python
# check duplication for specific unique combination columns 
cols = ['Title'] # Title cannot be a global key
duplicates = merged_df[merged_df.duplicated(cols,keep=False)]
print ("There are "+ str(len(duplicates)) + " duplicate records found")
duplicates.sort_values(by=cols).head(10)
```

## Global key 
The global key I propose is: 'Clothing ID', 'Price' and 'Title'. This unique combination is able to refer to a specific record within the dataframe. As you can see from the trial and error above with different combinations of variables, this combination is minimal and does not contain any duplicate records. Consideration was given to using Active Time instead of Price, active time is an observation of the time it has taken to leave the review and price is an attribute of the sale. It is possible the same item to be listed at the same price, but based on the data price is quite variable for items with the same clothing ID. Whilst the proposed global key works for the current dataset, it is POSSIBLE that a clothing item is sold for the same price, and a user/reviewer copies the title of another user's review, which would violate the key constraint. Without further information from the client/data provider, it is impossible to say whether or not the proposed global key is valid or not. The question that needs to be asked, is if there is a business rule/restriction on whether a 'Title' needs to be unique for each review for each item. It may be recommended that the client/data provider stores the information in the given dataframe in a relational database, and keeping an identification number not just for clothing items, but for individual reviews.
